```python
# Multiple-feature data

import numpy as np
import matplotlib.pyplot as plt

X = np.array([
    [1200, 3],
    [1800, 4],
    [2500, 5]
])

y = np.array([200000, 300000, 400000])
```

```python
# Prediction

def predict_multi(X, w, b):
    return X @ w + b
```

```python
# Test:

w = np.array([150, 20000])
b = 10000

predictions = predict_multi(X, w, b)
print("Predictions:", predictions)
```

```python
# Cost funtion

def compute_cost_multi(X, y, w, b):
    m = len(y)
    
    predictions = predict_multi(X, w, b)
    errors = predictions - y
    
    cost = (1 / (2 * m)) * np.sum(errors ** 2)
    
    return cost
```

```python
# Test:

cost = compute_cost_multi(X, y, w, b)

print("Cost:", cost)
```

```python
# Gradient

def compute_gradient_multi(X, y, w, b):
    m = len(y)
    
    predictions = predict_multi(X, w, b)
    errors = predictions - y
    
    dj_dw = (1 / m) * (X.T @ errors)
    dj_db = (1 / m) * np.sum(errors)
    
    return dj_dw, dj_db
```

```python
# Gradient descent

def gradient_descent_multi(X, y, w, b, alpha, num_iters):
    
    cost_history = []
    
    for i in range(num_iters):
        
        dj_dw, dj_db = compute_gradient_multi(X, y, w, b)
        
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        
        cost = compute_cost_multi(X, y, w, b)
        cost_history.append(cost)
        
    return w, b, cost_history
```

```python
# Train the model

w_init = np.array([150.0, 20000.0])
b_init = 10000.0

alpha = 1e-9
num_iters = 1000

w_final, b_final, cost_history = gradient_descent_multi(
    X,
    y,
    w_init,
    b_init,
    alpha,
    num_iters
)

print("Final w:", w_final)
print("Final b:", b_final)
```

```python
# Generate final predictions

predictions_final = predict_multi(
    X,
    w_final,
    b_final
)

print("Final predictions:", predictions_final)
```

```python
# Visualize cost

plt.plot(cost_history)

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration")

plt.show()
```
